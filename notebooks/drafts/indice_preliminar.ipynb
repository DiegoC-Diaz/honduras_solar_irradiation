### Construccion de un Indice Para Seleccionar una region

```python
file="../data/index/raw/inform_honduras_25_oct_2021.xlsx"


import pandas as pd

df_1 = pd.read_excel(
    file,
    sheet_name='COREL',
    skiprows=0,       # Salta las primeras 9 filas (para empezar a leer en la 10)
    usecols="A:AF",    # Lee únicamente el bloque de columnas de la C a la F
    nrows=299,          # A partir del inicio, lee solo 50 filas hacia abajo
    engine='calamine'
)

# Guardar solo ese bloque en CSV
df_1.to_csv('../data/index/processed/indicadores_riesgo_1.csv', index=False, encoding='utf-8')
```
